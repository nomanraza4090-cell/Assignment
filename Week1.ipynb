## 1 The Personal Assistant 

```python
# # Create a simple program that acts as a basic personal assistant by asking for user input and performing simple operations.

# Step 1: Ask for user's name and store it as a string
name = (input("What is your name? "))

# Step 2: Ask for birth year and store as integer
birth_year = int(input("What year were you born? "))

# Step 3: Ask for current age and store as integer
age = int(input("What is your current age? "))

# Step 4: Ask for current year and store as integer
current_year = int(input("Enter the current year: "))

# Step 5: Calculate the year they will turn 100 years old
year_turn_100 = current_year + (100 - age)

# Step 6: Ask for lucky number and store as float
number = float(input("Enter Your lucky number: "))

# Step 7: Print a summary message with all collected information
print("Hello," , name)
print("You were born in" , birth_year)
print("You are currently " , age , "years old.")
print("Your lucky number is: " , number)
print("You will turn 100 in the year: " ,year_turn_100)
```

## 2 The Shopping List Manager 

```python
#  Build a program that manages a shopping list using lists and dictionaries.

shopping_list = ["milk", "bread", "eggs", "apples", "chicken"]
item_prices = {"milk" : 120 , "bread" : 40 , "eggs" : 320 , "apple" :  400}


print("intial shopping list: " , shopping_list)
print("The third item is: " , shopping_list[2])


shopping_list.append("cheese")
shopping_list.append("Coffee")
print("Adding 'cheese' and 'coffee'...")


shopping_list.remove("chicken")
print("Removing 'chicken'...")


print("Updated Shopping List: " , shopping_list)


print("Item Prices: " , item_prices)
print("The price of apple is Rs" , item_prices["apple"])

print("---------Final Summary---------")
print("Items to buy" , shopping_list)
print("price lookup: ", item_prices)
```
